c. Is chemotherapy an effective treatment?

This graph does not necessarily tell us if chemotherapy is an effective treatment, only that patients who received it tend to have higher death proportions. This could mean many things. One interpretation is chemotherapy itself is associated with higher death rates, and some might jump to the conclusion that it 'causes' death. Another interpretation is that those receiving chemotherapy have a more aggressive form of cancer than those who did not undergo chemotherapy, and therefore the cancer severity rather than the chemotherapy is what is associated with higher death proportions. To determine if chemotherapy is an effective treatment, we must perform a paired experiment to measure the before and after of individual patients to see if the chemotherapy has improved their condition or not.
